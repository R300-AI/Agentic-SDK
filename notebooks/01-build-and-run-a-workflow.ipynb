# 做出一條可以跑的工作流程

這份教材先用最小但完整的例子，帶你從一個需求組出工作流程，送出問題，取得回覆。

## 在 Colab 準備環境

如果你是在 Colab 開啟，先複製專案並安裝套件；如果你已經在專案資料夾，可以直接跳過這格。

```python
from pathlib import Path
import os, sys, subprocess

if not Path('agentic_sdk').exists():
    if not Path('Agentic-SDK').exists():
        subprocess.run(['git', 'clone', 'https://github.com/R300-AI/Agentic-SDK.git'], check=True)
    os.chdir('Agentic-SDK')

subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-r', 'requirements.txt'], check=True)
print('Agentic SDK ready')
```

## 建立工作流程

這條流程會先保留使用者的問題，再從內建知識表找資料，最後直接回答。

```python
from agentic_sdk import Workflow
from agentic_sdk.modules import DirectAnswerAction, KeywordRetrieve, PassThroughPerceive

workflow = Workflow(
    workflow_name='入門問答 Agent',
    description='用內建資料回答常見問題。',
    perceive=PassThroughPerceive(),
    retrieve=KeywordRetrieve(items=[
        {'keywords': ['agentic sdk', 'sdk'], 'content': 'Agentic SDK 可以用工作流程組裝 Agent 行為。'},
        {'keywords': ['tsip'], 'content': 'TSiP 是工研院推動國產 AI 晶片落地的藍圖。'},
    ]),
    action=DirectAnswerAction(),
)

result = workflow.run('TSiP 是什麼？')
print(result.final_message)
```

## 看結果裡還有什麼

不要只看最後一句話。`result` 裡還有本次執行留下的資料，下一份教材會用它來追蹤每一步。

```python
print('session_id:', result.session_id)
print('visit_counts:', result.visit_counts)
print('entities:', result.entities)
```
