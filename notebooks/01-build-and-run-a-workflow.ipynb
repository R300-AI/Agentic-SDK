## 先載入這次會用到的東西這裡只載入四個角色：工作流程本體、輸入整理、查資料、回答。先知道每個角色要做什麼，後面再把它們接起來。

```python
from agentic_sdk import Workflowfrom agentic_sdk.modules import DirectAnswerAction, KeywordRetrieve, PassThroughPerceive
```

## 先準備 Agent 能查的資料這裡先用小型知識表。資料格式很單純：哪些關鍵字會命中，以及命中後要拿出哪段內容。

```python
knowledge_items = [    {'keywords': ['agentic sdk', 'sdk'], 'content': 'Agentic SDK 可以用工作流程組裝 Agent 行為。'},    {'keywords': ['tsip'], 'content': 'TSiP 是工研院推動國產 AI 晶片落地的藍圖。'},]knowledge_items
```

## 宣告輸入整理這一格只負責建立輸入整理模組。這個入門例子不改寫使用者文字，直接把原句交給後面使用。

```python
perceive = PassThroughPerceive()perceive
```

## 宣告查資料方式這一格把剛剛的知識表交給查資料模組。使用者問題進來後，會先用關鍵字找出最相關的內容。

```python
retrieve = KeywordRetrieve(items=knowledge_items, fallback='目前沒有找到相關資料。')retrieve
```

## 宣告回答方式這一格只建立回答模組。它會把查到的內容整理成最後回覆，不另外呼叫模型。

```python
action = DirectAnswerAction()action
```

## 最後才把模組接成工作流程前面每一格都只是準備零件；到這裡才把輸入整理、查資料、回答接成一條可以跑的工作流程。

```python
workflow = Workflow(    workflow_name='入門問答 Agent',    description='用內建資料回答常見問題。',    perceive=perceive,    retrieve=retrieve,    action=action,)workflow
```

## 送出第一個問題現在流程已經組好，可以送出一句使用者問題，看看最後回覆。

```python
result = workflow.run('TSiP 是什麼？')print(result.final_message)
```

## 看結果裡還有什麼不要只看最後一句話。`result` 裡還有本次執行留下的資料，後面的教材會用它來追蹤每一步。

```python
print('session_id:', result.session_id)print('visit_counts:', result.visit_counts)print('entities:', result.entities)
```
